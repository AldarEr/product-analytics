```python
import pandas as pd
```

```python
ab = pd.read_csv("../data/cookie_cats.csv")
print(len(ab))
print(ab['version'].value_counts())
ab.head()
```

```python
summary = ab.groupby('version')['retention_7'].agg(['mean','sum','count'])
summary
```

```python
r30 = ab.loc[ab["version"] == "gate_30", "retention_7"].mean()
r40 = ab.loc[ab["version"] == "gate_40", "retention_7"].mean()
print(f"gate_30 (контроль): {r30:.5f}  ({r30:.3%})")
print(f"gate_40 (вариант):  {r40:.4f}  ({r40:.2%})")
print(f"Разница: {(r40 - r30):.2%}")
```

```python
from statsmodels.stats.proportion import proportions_ztest
```

```python
returned = ab.groupby('version')['retention_7'].sum()
returned
```

```python
totals = ab.groupby('version')['retention_7'].count()
totals
```

```python
count = [returned["gate_30"], returned["gate_40"]]
nobs  = [totals["gate_30"],  totals["gate_40"]]
print(count)
print(nobs)
```

```python
z_stat, p_value = proportions_ztest(count,nobs)
```

```python
print(f"Вернулись:  gate_30 = {count[0]:,} из {nobs[0]:,}")
print(f"            gate_40 = {count[1]:,} из {nobs[1]:,}")
print(f"z-статистика: {z_stat:.3f}")
print(f"p-value:      {p_value:.4f}")
```

```python
ret1 = ab.groupby('version')['retention_1'].sum()
ret1
```

```python
tot1 = ab.groupby('version')['retention_1'].count()
tot1
```

```python
cou1 = [ret1['gate_30'],ret1['gate_40']]
nobs1 = [tot1['gate_30'],tot1['gate_40']]
```

```python
print(f"z-статистика: {z:.3f}")
print(f"p-value       {p:.4f}")
```

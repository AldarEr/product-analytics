```python
import pandas as pd
import requests
print("Библиотеки установлены")
```

```python
response = requests.get("https://dummyjson.com/products?limit=100")
print(response.status_code)
```

```python
data = response.json()
print(type(data))
print(data.keys())
```

```python
products = data["products"]
df = pd.DataFrame(products)
print(df.shape)
df.head(3)
```

```python
wanted = ['id','title','category','price','rating','stock']
products_df = df[wanted]

products_df.to_csv("../data/products_from_api.csv", index=False)
products_df.head()
```
